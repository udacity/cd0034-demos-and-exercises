## Classify Architecture Pain Points

A mid-size SaaS company, **CloudServe Inc.**, has the following data platform with an overall spend of **~$18,000/mo**:

### Current Architecture Details
- **Data Sources & Ingestion:**
  - **User Events:** Streaming clickstream events ingested through **Amazon Kinesis**. Logs land in S3, while team Lambdas write directly to Redshift.
  - **Billing Transactions:** Daily batch billing exports ingested via scheduled **AWS Glue** ETL jobs into S3.
- **Storage & Processing:**
  - **Amazon S3:** Raw event logs land here from Kinesis, but files are small (avg 200KB), unpartitioned, and in JSON format. Downstream Glue/Athena/Spark jobs must scan entire unpartitioned datasets with high request overhead.
  - **PostgreSQL (RDS):** Stores transactional application data containing customer information (PII) AND serves as an ad-hoc reporting database. Product teams and Tableau run analytical queries directly against production OLTP tables.
  - **Amazon Redshift:** A 2-node cluster ($6,400/mo) running 24/7 (with an estimated ~$2,500/mo in idle/off-peak waste). Multiple teams write directly to it via Lambda functions (small-block inserts). No sort keys or distribution keys are defined.
  - **AWS Glue:** 15 ETL jobs, 4 of which are deprecated but still scheduled (~$200+/mo wasted compute). No workflow dependency management exists between jobs.
- **Consumption:**
  - **Tableau:** Connected directly to Redshift (dashboard refreshes compete with Lambda/ETL writes for cluster resources) as well as PostgreSQL RDS.
- **Governance:**
  - **No data catalog or governance layer exists:** No centralized metadata, lineage, or fine-grained access control exists across services, creating compliance and PII exposure risks.
