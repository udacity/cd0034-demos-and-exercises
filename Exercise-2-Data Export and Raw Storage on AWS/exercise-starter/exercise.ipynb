## Design a Raw Landing Zone

**CloudServe Inc.** is redesigning their data ingestion. They have four data sources that need to land in S3 as their bronze layer:

| Source | Format | Frequency | Volume | Description / Source Details |
|--------|--------|-----------|--------|------------------------------|
| User clickstream events | JSON | Real-time (~2,000 events/sec) | ~50 GB/day | Ingested via streaming infrastructure |
| Billing transactions | CSV | Daily batch export at 2 AM UTC | ~500 MB/day | Exported from financial billing database |
| Support tickets (Zendesk API) | JSON | Hourly API poll | ~10,000 records/day | Extracted from external REST API |
| Product usage metrics | Parquet | Every 15 minutes from internal service | ~5 GB/day | Exported by internal microservice to an application S3 bucket |
